```python

# PROGRAM 3: ADULT INCOME - FEATURE ENGINEERING AND EDA

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 65)
print("ADULT INCOME - FEATURE ENGINEERING AND EDA PROJECT")
print("=" * 65)

# 2. Load Dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

columns = [
    "age", "workclass", "fnlwgt", "education", "education_num",
    "marital_status", "occupation", "relationship", "race", "sex",
    "capital_gain", "capital_loss", "hours_per_week",
    "native_country", "income"
]

df = pd.read_csv(
    url, names=columns, sep=",", skipinitialspace=True,
    na_values="?"
)

print("\n1. DATASET LOADED")
print("Dataset shape:", df.shape)
display(df.head())

# 3. Understand the Dataset
print("\n2. DATASET INFORMATION")
df.info()

print("\n3. STATISTICAL SUMMARY")
display(df.describe(include="all").T)

# 4. Check Missing Values
print("\n4. MISSING VALUES")
print(df.isnull().sum())

# 5. Clean Column Text
categorical_columns = df.select_dtypes(include="object").columns
for col in categorical_columns:
    df[col] = df[col].str.strip()

# 6. Handle Missing Values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

print("\n5. MISSING VALUES AFTER CLEANING")
print(df.isnull().sum().sum())

# 7. Exploratory Data Analysis
print("\n6. INCOME DISTRIBUTION")
print(df["income"].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="income")
plt.title("Income Category Distribution")
plt.xlabel("Income")
plt.ylabel("Number of People")
plt.show()

# Age Distribution
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x="age", bins=30, kde=True)
plt.title("Age Distribution")
plt.show()

# Education and Income
plt.figure(figsize=(12, 5))
sns.countplot(data=df, x="education", hue="income")
plt.title("Education Level vs Income")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Hours Worked Per Week
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="income", y="hours_per_week")
plt.title("Working Hours by Income Category")
plt.show()

# 8. Feature Engineering
df["age_group"] = pd.cut(
    df["age"],
    bins=[16, 25, 35, 45, 55, 65, 100],
    labels=["17-25", "26-35", "36-45", "46-55", "56-65", "65+"],
    right=True,
    include_lowest=True
)

df["is_married"] = df["marital_status"].isin(
    ["Married-civ-spouse", "Married-AF-spouse"]
).astype(int)

df["capital_net"] = df["capital_gain"] - df["capital_loss"]

df["works_long_hours"] = (
    df["hours_per_week"] > 40
).astype(int)

print("\n7. NEW FEATURES CREATED")
print(["age_group", "is_married", "capital_net", "works_long_hours"])
display(df[[
    "age", "age_group", "marital_status", "is_married",
    "capital_gain", "capital_loss", "capital_net",
    "hours_per_week", "works_long_hours", "income"
]].head(10))

# 9. Correlation Analysis
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 7))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Numerical Features")
plt.show()

# 10. Save Cleaned and Engineered Dataset
output_file = "adult_income_feature_engineered.csv"
df.to_csv(output_file, index=False)

print("\n8. FINAL DATASET")
print("Final shape:", df.shape)
print("Saved file:", output_file)
display(df.head())

# 11. Conclusion
print("\nCONCLUSION")
print("We explored the Adult Income dataset, handled missing values,")
print("visualized important patterns, and created new features.")
print("The engineered dataset is saved for future machine learning.")
```
