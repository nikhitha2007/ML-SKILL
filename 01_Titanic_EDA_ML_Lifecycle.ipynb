```python

# PROGRAM 1: TITANIC SURVIVAL - EDA AND ML LIFECYCLE MAPPING

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 60)
print("TITANIC SURVIVAL - EDA AND ML LIFECYCLE MAPPING")
print("=" * 60)

# 2. Load Dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("\n1. DATASET LOADED SUCCESSFULLY")
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# 3. Dataset Information
print("\n2. DATASET INFORMATION")
df.info()

# 4. Statistical Summary
print("\n3. STATISTICAL SUMMARY")
display(df.describe())

# 5. Check Missing Values
print("\n4. MISSING VALUES")
print(df.isnull().sum())

# 6. Survival Counts
print("\n5. SURVIVAL COUNTS")
print("0 = Did not survive, 1 = Survived")
print(df["Survived"].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Survived")
plt.title("Titanic Passenger Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()

# 7. Survival by Gender
print("\n6. SURVIVAL BY GENDER")
display(pd.crosstab(df["Sex"], df["Survived"]))

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Survival by Gender")
plt.show()

# 8. Survival by Passenger Class
print("\n7. SURVIVAL BY PASSENGER CLASS")
display(pd.crosstab(df["Pclass"], df["Survived"]))

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Pclass", hue="Survived")
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.show()

# 9. Age Distribution
print("\n8. AGE DISTRIBUTION BY SURVIVAL")
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", hue="Survived",
             bins=25, kde=True, element="step")
plt.title("Age Distribution by Survival")
plt.show()

# 10. Correlation Heatmap
print("\n9. CORRELATION BETWEEN NUMERICAL FEATURES")
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(9, 6))
sns.heatmap(numeric_df.corr(), annot=True,
            cmap="coolwarm", fmt=".2f")
plt.title("Titanic Dataset Correlation Heatmap")
plt.show()

# 11. Machine Learning Lifecycle
print("\n10. MACHINE LEARNING LIFECYCLE")
lifecycle = [
    "1. Problem Definition",
    "2. Data Collection",
    "3. Exploratory Data Analysis",
    "4. Data Preprocessing",
    "5. Feature Engineering",
    "6. Model Selection",
    "7. Model Training",
    "8. Model Evaluation",
    "9. Deployment and Monitoring"
]

for stage in lifecycle:
    print(stage)

# 12. Conclusion
print("\n" + "=" * 60)
print("CONCLUSION")
print("=" * 60)
print("We successfully explored the Titanic dataset, examined")
print("missing values and statistical summaries, visualized")
print("survival patterns, and reviewed the ML lifecycle.")
print("This project builds a foundation for further ML programs.")
```
